# Tandem lift – loss of load: coupled ship/load simulation

Python implementation of the simulation method of **Hatecke (2016), *Dynamics of Loss of Load during Tandem Lifts*** (TUHH dissertation, Ch. 2–4):

* a crane ship and the lifted load are two rigid bodies connected by the **one remaining crane rope** (rigid, massless rod) → **11 DOF** in minimal coordinates (ship 6, rope 2, load 3) – eqs. (2.2)–(2.16);
* equations of motion via Jacobian + **D'Alembert's principle** (2.44)–(2.48);
* **failure of the second rope**: switch to 12 independent DOF and state transformation (2.49)–(2.69);
* applied forces: gravity, **non‑linear hydrostatics**, **radiation forces as a state‑space system** (no convolution), **viscous roll damping**;
* analyses of Ch. 4: delay time `t_d`, crane configuration (load yaw `e0`), quasi‑static criterion adjustment, direct assessment of required GM.

The test case is the model‑test ship of Ch. 3 (Tables 3.1–3.4, full scale).

### What is taken from the thesis and what is *my* simplification
| Item | In the thesis | In this notebook |
|---|---|---|
| Multibody kinematics / kinetics, rope failure, state transformation | as in Ch. 2 | **same**; $\dot J\dot y$ by a directional finite difference instead of eq. (2.29); load Euler sequence changed to $R_z R_x R_y$ (gimbal lock at 90° load swing with the thesis' $R_zR_yR_x$) |
| Hydrostatics | 3‑DOF look‑up table from hull offsets | lever‑arm curve **digitised from Fig. 3.4** (sine‑series fit) for roll; linear heave (`Awp`) and pitch (`GM_L`) – assumed values |
| Radiation forces | BEM (NEWDRIFT) + own MIMO time‑domain identification with passivity enforcement | **placeholder** diagonal `A∞`, `B(ω)` (assumed), retardation function `K(τ)` (2.78) fitted by a stable state‑space model; stability + passivity are *checked* |
| Viscous roll damping | Ikeda simplified method | linear equivalent damping ratio, calibrated on the first roll peak of Fig. 3.14 |
| After 2nd rope failure | 12 DOF | ship (6 DOF) only – the free load is decoupled from the ship (eq. 2.63) |

So the numbers are *consistent with* – not a replacement of – the validated thesis results. Plug in your own BEM data (`A_inf`, `B(ω)`) to go further.

### Two‑vessel extension (this version)
All ship‑specific data now live in a **`Vessel`** object (Sect. 1b: mass/inertia, hydrostatics, hydrodynamic placeholders, ballast), and the rope/load geometry in **`Config(V, …)`**.  
Sections 1–10 are the unchanged single‑ship study of the thesis (they run through the same code with `V_thesis` and reproduce the previous numbers). **Sect. 11** adds the case of **two vessels**, one rope each: after the first rope fails, the *remaining vessel* + load is the 11‑DOF system of this notebook, the *released vessel* is decoupled and simply unloaded.

## 1b  Vessel object and lifting condition (new – needed for two vessels)

Everything that belongs to **one ship** is bundled in `Vessel`: mass and inertia, hydrostatics (`GZ_hull`, `Awp`, `GM_L`), hydrodynamic placeholders (`A_inf`, used by `build_radiation`) and the **loading condition before the first rope fails**.

`Vessel.design_ballast(share, pre_point, F_rope)` computes the COG offsets `x_G`, `e_G` such that the vessel floats **upright and untrimmed under the force of its own rope** (the thesis does the same for the model-test ship with the full load, Tab. 3.3):

$$e_G=\frac{\text{share}\,m_L\,y_{pre}+z_{rel}F_y/g}{m_S},\qquad x_G=\frac{x_B\,(m_S+\text{share}\,m_L)-x_{pre}\,\text{share}\,m_L-z_{rel}F_x/g}{m_S}$$

($F$ = rope force *on the ship*, $z_{rel}$ = height of the force point above the ship COG). A horizontal rope force (inclined rope) is held by a **constant station‑keeping force `F_sk` and yaw moment `M_sk_z`**; they do not react to the rope force after the failure. The vessel's displacement before the failure is $m_S+\text{share}\,m_L$; after the failure it suddenly carries the whole load.

For `share = 1` (both cranes on the same ship) this reproduces exactly the loading condition of the thesis ship.

```python
class Vessel:
    """Everything that belongs to ONE ship: mass/inertia, hydrostatics, hydrodynamic placeholders and its loading
    condition (ballast) before the first rope fails.  Any number of vessels can exist at the same time."""
    def __init__(s,name='thesis ship',m_S=m_S,KG_S=KG_S,k=(k_xx,k_yy,k_zz),x_B=x_B,Awp=Awp,GM_L=GM_L,
                 KM=GM0+KG_tot,a_gz=a_gz,phi_f=phi_f,draught_full=draught):
        s.name,s.m_S,s.KG_S,s.x_B,s.Awp,s.GM_L,s.KM,s.phi_f,s.draught_full=name,m_S,KG_S,x_B,Awp,GM_L,KM,phi_f,draught_full
        s.a_gz=np.array(a_gz); s.k_xx,s.k_yy,s.k_zz=k
        s.I_S=m_S*np.diag(np.square(k))
        s.A_inf=np.array([0.05*m_S,0.9*m_S,1.1*m_S,0.15*m_S*s.k_xx**2,1.0*m_S*s.k_yy**2,0.8*m_S*s.k_zz**2])  # placeholders
        s.share=None; s.x_G=s.e_G=None; s.F_sk=np.zeros(3); s.M_sk_z=0.0
    r_G=property(lambda s: np.array([s.x_G,-s.e_G,s.KG_S]))
    w_phi=property(lambda s: np.sqrt(g*s.m_pre*s.GM_pre/(s.m_S*s.k_xx**2*1.15)))
    def GZ(s,phi,dGM=0.0):
        return np.sin(np.outer(np.atleast_1d(phi),_K))@s.a_gz+dGM*np.sin(phi)
    def h2(s,phi):                                   # lever arm after loss of the WHOLE load
        return s.GZ(phi)[0]-s.e_G*np.cos(phi)
    def design_ballast(s,share,pre_point,F_rope,m_load=m_L):
        """x_G, e_G (+ constant station-keeping wrench) so that the vessel floats upright and untrimmed under the rope
        force F_rope (force ON the ship, N) acting at pre_point (ship coordinates) before the first failure."""
        pre_point=np.asarray(pre_point,float); Fx,Fy,_=F_rope; z_rel=pre_point[2]-s.KG_S
        s.share=share; s.m_pre=s.m_S+share*m_load                       # displacement before the first failure
        s.e_G=(share*m_load*pre_point[1]+z_rel*Fy/g)/s.m_S
        s.x_G=(s.x_B*s.m_pre-pre_point[0]*share*m_load-z_rel*Fx/g)/s.m_S
        rel=pre_point-s.r_G
        s.F_sk=-np.array([Fx,Fy,0.0]); s.M_sk_z=-(rel[0]*Fy-rel[1]*Fx)   # holds the horizontal rope force before the failure
        s.draught_pre=s.draught_full-(1-share)*m_load/(rho*s.Awp)
        s.GM_pre=s.KM-(s.m_S*s.KG_S+share*m_load*pre_point[2])/s.m_pre    # GM with the load hanging in the crane tip
        s.z0=s.KG_S-s.draught_pre                                        # COG height above the waterline
        return s

# thesis ship: both cranes on the ship, ballasted for the full load (thesis Tab. 3.3)
V_thesis=Vessel('thesis ship').design_ballast(1.0,np.array([com_L[0],tip_fwd[1],tip_fwd[2]]),np.array([0,0,-m_L*g]))
print(f"V_thesis: x_G = {V_thesis.x_G:.2f} m (global {x_G:.2f}),  e_G = {V_thesis.e_G:.3f} m (global {e_G:.3f}),  GM = {V_thesis.GM_pre:.2f} m (thesis {GM0})")
assert abs(V_thesis.x_G-x_G)<1e-9 and abs(V_thesis.e_G-e_G)<1e-9
```

## 2  Kinematics, Jacobian (eqs. 2.2–2.26)

Generalised coordinates $y=[\xi_S,\alpha_S,\alpha_R,\alpha_H]^T$ (11 DOF). The Jacobian maps $\dot y$ to $[\dot r_{ES},\omega_{ES},\dot r_{EL},\omega_{EL}]$. It is verified against finite differences of the load position and orientation.

```python
def Rx(a):
    c,s=np.cos(a),np.sin(a); return np.array([[1,0,0],[0,c,-s],[0,s,c]])
def Ry(a):
    c,s=np.cos(a),np.sin(a); return np.array([[c,0,s],[0,1,0],[-s,0,c]])
def Rz(a):
    c,s=np.cos(a),np.sin(a); return np.array([[c,-s,0],[s,c,0],[0,0,1]])
def hat(v): return np.array([[0,-v[2],v[1]],[v[2],0,-v[0]],[-v[1],v[0],0]])
def P_zyx(th,psi):
    return np.array([[np.cos(th)*np.cos(psi),-np.sin(psi),0],[np.cos(th)*np.sin(psi),np.cos(psi),0],[-np.sin(th),0,1]])
def eulzyx(R):
    return np.array([np.arctan2(R[2,1],R[2,2]),-np.arcsin(np.clip(R[2,0],-1,1)),np.arctan2(R[1,0],R[0,0])])

def eul_zxy(R):
    """angles (ep,ga,de) with R = Rz(ep) Rx(ga) Ry(de)  (load Euler sequence of this notebook)"""
    ga=np.arcsin(np.clip(R[2,1],-1,1)); de=np.arctan2(-R[2,0],R[2,2])
    Rzm=R@(Rx(ga)@Ry(de)).T; ep=np.arctan2(Rzm[1,0],Rzm[0,0])
    return ep,ga,de

class Config:
    """Geometry of the REMAINING rope on vessel V after the first rope has failed.
    e0    : load yaw; the remaining lifting eye sits at Rz(e0)(ex,0,ez) from the load COG
    share : fraction of the load carried by this vessel's rope BEFORE the first failure (1 = both cranes on this ship)
    theta : rope inclination from the vertical; >0: crane tip outboard of the eye (rope leans towards the load)
    tip   : crane tip in ship coordinates; None -> load COG fixed at com_L and the tip follows (thesis, Sect. 4.3)
    design: True -> the ballast of V is designed for this lift (two vessels, crane tip fixed on the ship)"""
    def __init__(s,V,e0=np.pi,share=1.0,theta=0.0,tip=None,design=False,com_L=com_L,ex=ex,ez=ez,l_rope=l_rope,m_L=m_L,I_L=I_L):
        s.V,s.e0,s.share,s.theta=V,e0,share,theta
        s.l_rope,s.m_L,s.I_L,s.ex,s.ez=l_rope,m_L,I_L,ex,ez
        s.eye=Rz(e0)@np.array([ex,0,ez]); n=s.eye[:2]/np.hypot(*s.eye[:2])      # eye relative to the load COG, outboard direction
        s.u=np.array([-np.sin(theta)*n[0],-np.sin(theta)*n[1],-np.cos(theta)])   # rope direction tip -> hook
        s.tip=com_L+s.eye-l_rope*s.u if tip is None else np.asarray(tip,float)
        s.T_pre=share*m_L*g/(-s.u[2])                                            # rope tension before the first failure
        if design: V.design_ballast(share,s.tip,s.T_pre*s.u,m_L)
        s.rSR_S=s.tip-V.r_G; s.rHL_L=-np.array([ex,0,ez])
        al=np.arcsin(s.u[1]); be=np.arctan2(-s.u[0],-s.u[2])                      # rope angles (ship upright)
        ep,ga,de=eul_zxy((Ry(be)@Rx(al)).T@Rz(e0))                                # load hangs level, yawed by e0
        s.y0=np.zeros(11); s.y0[6:8]=[al,be]; s.y0[8:11]=[ga,de,ep]               # [xi(3),phi,th,psi,alpha,beta,gamma,delta,e]
        s.y0[2]=V.z0                                                              # E-frame: z of the ship COG above the waterline

def kin(y,cfg):
    phi,th,psi=y[3:6]; al,be=y[6:8]; ga,de,ep=y[8:11]
    RES=Rz(psi)@Ry(th)@Rx(phi); RSR=Ry(be)@Rx(al); RRL=Rz(ep)@Rx(ga)@Ry(de)
    RER=RES@RSR
    r_SR=RES@cfg.rSR_S; r_RH=RER@np.array([0,0,-cfg.l_rope]); r_HL=RER@RRL@cfg.rHL_L
    return RES,RSR,RRL,RER,r_SR,r_RH,r_HL

def J11(y,cfg):
    RES,RSR,RRL,RER,r_SR,r_RH,r_HL=kin(y,cfg)
    be=y[7]; ga,de,ep=y[8:11]
    PES=P_zyx(y[4],y[5])
    PSR=RES@np.array([[np.cos(be),0],[0,1],[-np.sin(be),0]])
    c1=np.array([np.cos(ep),np.sin(ep),0])
    c2=np.array([-np.sin(ep)*np.cos(ga),np.cos(ep)*np.cos(ga),np.sin(ga)])
    PRH=RER@np.column_stack([c1,c2,[0,0,1]])
    r_RL=r_RH+r_HL; r_SL=r_SR+r_RL
    J=np.zeros((12,11))
    J[0:3,0:3]=np.eye(3); J[3:6,3:6]=PES
    J[6:9,0:3]=np.eye(3); J[6:9,3:6]=-hat(r_SL)@PES; J[6:9,6:8]=-hat(r_RL)@PSR; J[6:9,8:11]=-hat(r_HL)@PRH
    J[9:12,3:6]=PES; J[9:12,6:8]=PSR; J[9:12,8:11]=PRH
    return J
def Jdot_yd(Jfun,y,yd,cfg):
    eps=1e-6/max(1.0,np.abs(yd).max())
    return (Jfun(y+eps*yd,cfg)-Jfun(y-eps*yd,cfg))@yd/(2*eps)

def r_load(y,cfg):
    RES,RSR,RRL,RER,r_SR,r_RH,r_HL=kin(y,cfg); return y[0:3]+r_SR+r_RH+r_HL
def R_load(y,cfg):
    RES,RSR,RRL,RER,*_=kin(y,cfg); return RER@RRL
```

```python
def jac_check(cfg,seed=0):
    rng=np.random.default_rng(seed); y=cfg.y0+0.2*rng.standard_normal(11); yd=rng.standard_normal(11)
    v=J11(y,cfg)@yd; eps=1e-6
    dr=(r_load(y+eps*yd,cfg)-r_load(y-eps*yd,cfg))/(2*eps)
    Rd=(R_load(y+eps*yd,cfg)-R_load(y-eps*yd,cfg))/(2*eps); W=Rd@R_load(y,cfg).T
    print("  max error load velocity   :",np.abs(dr-v[6:9]).max())
    print("  max error load ang. vel.  :",np.abs(np.array([W[2,1],W[0,2],W[1,0]])-v[9:12]).max())

cfg=Config(V_thesis,np.pi)    # e0 = 180°: model-test configuration (aft hook released first, forward crane remains)
print("thesis configuration (vertical rope):"); jac_check(cfg)
print("  remaining crane tip (x,y,z) =",cfg.tip.round(2))
# same check for an INCLINED rope (two-vessel geometry: crane tip fixed on the ship, rope leaning 8° towards the load)
cfg_t=Config(Vessel('test'),1.5*np.pi,share=0.5,theta=np.deg2rad(8),design=True)
print("inclined rope, 8°, share 0.5:"); jac_check(cfg_t,1)
print("  initial rope angles alpha, beta =",np.rad2deg(cfg_t.y0[6:8]).round(2),"°;  load level before the failure:",np.allclose(R_load(cfg_t.y0,cfg_t),Rz(cfg_t.e0)))
print("  load COG relative to ship COG = com_L - r_G :",np.allclose(r_load(cfg_t.y0,cfg_t)-cfg_t.y0[:3],com_L-cfg_t.V.r_G))
```

## 3  Radiation forces as a state‑space system (Sect. 2.5)

Cummins' equation (2.76) with the convolution $\int K(\tau)\dot\xi(t-\tau)d\tau$ is replaced by $\dot x=Ax+B\dot\xi,\ \mu=Cx$ (2.80). The retardation function $K(\tau)=\frac2\pi\int_0^\infty[B(\omega)-B_\infty]\cos\omega\tau\,d\omega$ (2.78) is fitted with damped sinusoids (variable projection) and realised as a real block‑diagonal state‑space model (3 oscillators → order 6 per DOF).  
Requirements (2.5.1): **stability** (all eigenvalues of `A` in the left half plane) and **passivity** ($\Re\,\hat H(i\omega)\ge0$, eq. 2.81–2.83) – both are checked below.

> `A_inf` and the shape of `B(ω)` are **placeholders**. Replace them with BEM output to reproduce a specific ship.

```python
def B_shape(w,Bpk,wp):
    x=w/wp; return Bpk*x**2*np.exp(1-x**2)
def retardation(Bpk,wp,tau):
    w=np.linspace(0,6*max(wp,0.3)+3,6000)
    Bw=B_shape(w,Bpk,wp)
    return np.array([2/np.pi*np.trapezoid(Bw*np.cos(w*t),w) for t in tau])
def fit_ss(tau,K,npairs=3):
    def lin(p):
        a=np.exp(p[:npairs]); b=p[npairs:]
        cols=[]
        for ai,bi in zip(a,b):
            cols+= [np.exp(-ai*tau)*np.cos(bi*tau),np.exp(-ai*tau)*np.sin(bi*tau)]
        return np.column_stack(cols)
    def res(p):
        M=lin(p); c=np.linalg.lstsq(M,K,rcond=None)[0]; return M@c-K
    best=None
    for trial in range(8):
        rng=np.random.default_rng(trial)
        p0=np.r_[np.log(rng.uniform(0.1,0.6,npairs)),rng.uniform(0.2,1.6,npairs)]
        r=least_squares(res,p0)
        if best is None or r.cost<best.cost: best=r
    p=best.x; M=lin(p); c=np.linalg.lstsq(M,K,rcond=None)[0]
    n=2*npairs; A=np.zeros((n,n)); B=np.zeros(n); C=np.zeros(n)
    for i in range(npairs):
        a=np.exp(p[i]); b=p[npairs+i]
        A[2*i:2*i+2,2*i:2*i+2]=[[-a,b],[-b,-a]]; B[2*i]=1; C[2*i]=c[2*i]; C[2*i+1]=-c[2*i+1]
    return A,B,C
def Bhat(A,B,C,w):
    n=len(B); return np.array([ (C@np.linalg.solve(1j*wi*np.eye(n)-A,B)).real for wi in w])

# 6-DOF diagonal hydrodynamic data of ONE vessel (placeholders in Vessel.A_inf / below: replace by BEM results)
def build_radiation(V,zeta_rad=0.04,npairs=3,verbose=False):
    m_S,k_xx,k_yy,k_zz,A_inf=V.m_S,V.k_xx,V.k_yy,V.k_zz,V.A_inf
    C33=rho*g*V.Awp; C44=V.m_pre*g*V.GM_pre; C55=V.m_pre*g*V.GM_L
    wp=np.array([0.3,0.4,0.85,0.55,0.8,0.4])
    crit=lambda C,M:2*np.sqrt(C*M)
    Bpk=np.array([0.05*(m_S*1.05)*0.3, 0.08*(m_S*1.9)*0.4, 0.2*crit(C33,m_S*2.1),
                  zeta_rad*crit(C44,m_S*k_xx**2*1.15),0.2*crit(C55,m_S*k_yy**2*2.0),0.05*(A_inf[5]+m_S*k_zz**2)*0.4])
    tau=np.arange(0,120,0.25); blocks=[]; fits=[]
    for i in range(6):
        K=retardation(Bpk[i],wp[i],tau); A,B,C=fit_ss(tau,K,npairs); blocks.append((A,B,C)); fits.append((Bpk[i],wp[i]))
    n=2*npairs
    Aall=np.zeros((6*n,6*n)); Ball=np.zeros((6*n,6)); Call=np.zeros((6,6*n))
    for i,(A,B,C) in enumerate(blocks):
        Aall[i*n:(i+1)*n,i*n:(i+1)*n]=A; Ball[i*n:(i+1)*n,i]=B; Call[i,i*n:(i+1)*n]=C
    return Aall,Ball,Call,blocks,fits
```

```python
Aall,Ball,Call,blocks,fits=build_radiation(V_thesis)
w=np.linspace(0,3,300); labels=['surge','sway','heave','roll','pitch','yaw']
fig,axs=plt.subplots(1,2,figsize=(10,3.4))
for i,c in zip([3,2],['C0','C1']):
    A,B,C=blocks[i]; tgt=B_shape(w,*fits[i]); got=Bhat(A,B,C,w)
    axs[0].plot(w,tgt/1e6,c+'-',label=labels[i]+' target'); axs[0].plot(w,got/1e6,c+'--',label=labels[i]+' state-space')
axs[0].set(xlabel='ω [rad/s]',ylabel='B(ω)−B∞  [MN·m·s/rad or MN·s/m]',title='Added damping'); axs[0].legend(fontsize=7)
tau=np.arange(0,120,.25); K=retardation(*fits[3],tau)
A,B,C=blocks[3]; Kss=[C@expm(A*t)@B for t in tau[:200]]
axs[1].plot(tau[:200],K[:200]/1e6,'k-',label='K(τ) from B(ω)'); axs[1].plot(tau[:200],Kss,'r--',label='state-space'); axs[1].set(xlabel='τ [s]',title='Retardation function (roll)'); axs[1].legend()
plt.tight_layout(); plt.show()
for i,(A,B,C) in enumerate(blocks):
    print(f"{labels[i]:6s}: max Re(eig A) = {np.linalg.eigvals(A).real.max():7.3f}   min B̆(ω)/max B̆ = {Bhat(A,B,C,w).min()/Bhat(A,B,C,w).max():.1e}  (passive if ≥ 0 within fit tolerance)")
```

## 4  Equations of motion (Sect. 2.2–2.3)

Phase 1 (rope still connected, 11 DOF):  
$J^TM_tJ\,\ddot y = J^T(q_{t,a}-M_t\dot J\dot y-k_t)$, eq. (2.44).  
Applied forces $q_{t,a}$ (2.70): gravity, hydrostatic force/moment, radiation wrench $-\mu$, viscous roll moment $n_v=-d_{44}\,p$ (2.115). Added mass `A∞` is added to the ship mass matrix.

Hydrostatic wrench (replacing the 3‑DOF look‑up table): $F_B=\rho g\nabla_0-\rho g A_{wp}\zeta$, roll moment $-F_B\,[GZ_{hull}(\varphi)-e_G\cos\varphi]$, pitch moment $-F_B[(x_B-x_G)\cos\vartheta+GM_L\sin\vartheta]$.

Phase 2 (second rope failed): the load is a free body and no longer influences the ship (2.63); the ship (6 DOF) is integrated with the same forces and the radiation states carried over. The load state at the instant of failure follows from eqs. (2.64)–(2.69) (`load_state_transform`).

*Two‑vessel version:* every ship‑specific quantity is read from `h.V` (the `Vessel`), every load/rope quantity from `cfg`. The hydrostatic force is referenced to the displacement **before** the first failure, `V.m_pre = m_S + share·m_L`; a constant station‑keeping wrench `V.F_sk`, `V.M_sk_z` holds the horizontal rope force of an inclined rope. For `share = 1` all of this vanishes and the equations are those of the thesis.

```python
class Hydro: pass
def make_hydro(V,zeta_rad=0.04,zeta_visc=0.11,dGM=0.0,npairs=3):
    h=Hydro(); h.V=V; h.A,h.B,h.C,*_=build_radiation(V,zeta_rad,npairs)
    h.d44=2*zeta_visc*V.w_phi*(V.m_S*V.k_xx**2+V.A_inf[3]); h.dGM=dGM; h.n=h.A.shape[0]
    return h

def ship_mats(RES,V):
    Mtr=RES@np.diag(V.m_S+V.A_inf[:3])@RES.T
    Irot=RES@(V.I_S+np.diag(V.A_inf[3:]))@RES.T
    return Mtr,Irot
def ship_forces(yS,ydS,x,h,z0):
    V=h.V
    phi,th,psi=yS[3:6]; RES=Rz(psi)@Ry(th)@Rx(phi)
    wE=P_zyx(th,psi)@ydS[3:6]
    nu=np.r_[RES.T@ydS[0:3],RES.T@wE]
    mu=h.C@x
    FB=V.m_pre*g-rho*g*V.Awp*(yS[2]-z0)                 # buoyancy of the vessel; reference = displacement before the first failure
    Mx=-FB*(V.GZ(phi,h.dGM)[0]-V.e_G*np.cos(phi))
    My=-FB*((V.x_B-V.x_G)*np.cos(th)+V.GM_L*np.sin(th))
    f=np.array([0,0,FB-V.m_S*g])+RES@(-mu[0:3])+V.F_sk   # F_sk: constant station-keeping force (zero for vertical ropes)
    n=Rz(psi)@np.array([Mx,My,0])+RES@(-mu[3:6]+np.array([-h.d44*nu[3],0,0]))+np.array([0,0,V.M_sk_z])
    return f,n,nu,RES

def rhs1(t,s,cfg,h,z0):
    V=h.V
    y=s[:11]; yd=s[11:22]; x=s[22:]
    f,n,nu,RES=ship_forces(y[:6],yd[:6],x,h,z0)
    J=J11(y,cfg); Jd=Jdot_yd(J11,y,yd,cfg)
    Mtr,Irot=ship_mats(RES,V)
    RELf=kin(y,cfg); RL=RELf[3]@RELf[2]
    IL=RL@cfg.I_L@RL.T
    Mt=np.zeros((12,12)); Mt[0:3,0:3]=Mtr; Mt[3:6,3:6]=Irot; Mt[6:9,6:9]=cfg.m_L*np.eye(3); Mt[9:12,9:12]=IL
    v=J@yd; wS=v[3:6]; wL=v[9:12]
    k=np.r_[0,0,0,np.cross(wS,Irot@wS),0,0,0,np.cross(wL,IL@wL)]
    qa=np.r_[f,n,[0,0,-cfg.m_L*g],[0,0,0]]
    ydd=np.linalg.solve(J.T@Mt@J,J.T@(qa-Mt@Jd-k))
    return np.r_[yd,ydd,h.A@x+h.B@nu]

def constraint_force(s,cfg,h,z0):
    y=s[:11]; yd=s[11:22]; x=s[22:]
    d=rhs1(0,s,cfg,h,z0); ydd=d[11:22]
    J=J11(y,cfg); Jd=Jdot_yd(J11,y,yd,cfg)
    a=J@ydd+Jd
    RELf=kin(y,cfg); RL=RELf[3]@RELf[2]; IL=RL@cfg.I_L@RL.T
    wL=(J@yd)[9:12]
    fL=cfg.m_L*a[6:9]-np.array([0,0,-cfg.m_L*g])
    return fL

def rhs2(t,s,h,z0):
    V=h.V
    y=s[:6]; yd=s[6:12]; x=s[12:]
    f,n,nu,RES=ship_forces(y,yd,x,h,z0)
    Mtr,Irot=ship_mats(RES,V)
    P=P_zyx(y[4],y[5]); e=1e-6
    Pd=((P_zyx(y[4]+e*yd[4],y[5]+e*yd[5])-P_zyx(y[4]-e*yd[4],y[5]-e*yd[5]))/(2*e))
    w=P@yd[3:6]
    acc_t=np.linalg.solve(Mtr,f)
    wd=np.linalg.solve(Irot,n-np.cross(w,Irot@w))
    ang=np.linalg.solve(P,wd-Pd@yd[3:6])
    return np.r_[yd,acc_t,ang,h.A@x+h.B@nu]

def simulate(td,cfg,h,t_end_after=25.0,rtol=1e-7,atol=1e-9,dense1=None):
    z0=cfg.y0[2]
    if dense1 is None:
        dense1=run_phase1(cfg,h,td+1.0,rtol,atol)
    s=dense1.sol(td)
    yS=np.r_[s[0:6],s[11:17]]
    s2=np.r_[yS,s[22:]]
    sol2=solve_ivp(rhs2,[td,td+t_end_after],s2,args=(h,z0),method='DOP853',rtol=rtol,atol=atol,max_step=0.5,dense_output=True)
    return dense1,sol2,s
def run_phase1(cfg,h,t_end,rtol=1e-7,atol=1e-9):
    s0=np.r_[cfg.y0,np.zeros(11),np.zeros(h.n)]
    return solve_ivp(rhs1,[0,t_end],s0,args=(cfg,h,cfg.y0[2]),method='DOP853',rtol=rtol,atol=atol,max_step=0.25,dense_output=True)
def run_release(cfg,h,t_end,rtol=1e-7,atol=1e-9):
    """RELEASED vessel (its rope has failed at t = 0): the rope force vanishes, the ship (6 DOF, rhs2) is integrated
    from its pre-failure equilibrium.  Needs no coupling to the other vessel."""
    s0=np.r_[cfg.y0[:6],np.zeros(6),np.zeros(h.n)]
    return solve_ivp(rhs2,[0,t_end],s0,args=(h,cfg.y0[2]),method='DOP853',rtol=rtol,atol=atol,max_step=0.5,dense_output=True)

def load_state_transform(s,cfg):
    y=s[:11]; yd=s[11:22]; J=J11(y,cfg); v=J@yd
    xiL=r_load(y,cfg); xiLd=v[6:9]; RL=R_load(y,cfg); aL=eulzyx(RL)
    return xiL,xiLd,aL,np.linalg.solve(P_zyx(aL[1],aL[2]),v[9:12])
```

## 9  Quasi‑static adjustment of the area‑ratio criterion (Sect. 4.4)

If the unexcited ship oscillates about the new heel `φc` (static heel with the full load on the remaining tip) starting from `φ1`, the most dangerous roll angle at which the second crane can fail is the extremum of that oscillation, `φx = 2φc − φ1` (cf. eqs. 4.1–4.2). Replacing `φ1` by `φx` in the area ratio `A2/A1` (Fig. 1.5) accounts for the worst failure order and timing. Below, the criterion is evaluated for the test lift and for the two extreme crane configurations. The dynamic simulation above is the reference.

```python
def static_heel(y_tip):
    # equilibrium of ship (23300 t displacement) with the full load hanging in the crane tip at transverse position y_tip
    z_t=tip_fwd[2]-KG_S
    def M(p):
        arm=(y_tip+e_G)*np.cos(p)-z_t*np.sin(p)
        return -m_tot*g*(GZ_hull(p)[0]-e_G*np.cos(p))-m_L*g*arm
    return brentq(M,-0.6,0.6)
def area_ratio(phi1):
    p2=brentq(h2,0.1,1.0); a=np.linspace(phi1,p2,500); b=np.linspace(p2,phi_f,500)
    A1=np.trapezoid(-np.array([h2(p) for p in a]),a); A2=np.trapezoid([h2(p) for p in b],b); return A1,A2
phi1=static_heel(tip_fwd[1]); print(f"initial heel with the load in both tips: {np.rad2deg(phi1):.2f}°")
rows=[]
for name,yt in [('test lift (no transverse offset)',tip_fwd[1]),('outboard crane remains (e0=90°)',tip_fwd[1]+ex),('inboard crane remains (e0=270°)',tip_fwd[1]-ex)]:
    phic=static_heel(yt); phix=2*phic-phi1; phiw=min(phi1,phix)      # worst extreme of the oscillation (towards the load side)
    A1,A2=area_ratio(phi1); A1w,A2w=area_ratio(phiw)
    rows.append((name,np.rad2deg(phic),np.rad2deg(phiw),A2/A1,A2w/A1w))
print(pd.DataFrame(rows,columns=['configuration','φc [°]','worst φ at failure [°]','A2/A1 (φ1)','A2/A1 (adjusted)']).round(2).to_string(index=False))
print("\nAn energy balance without dissipation lets the roll reach φ_f =",np.rad2deg(phi_f).round(1),"° when A2/A1 ≈ 1; the simulation of the test lift gives at most",R[:,1].max().round(1),"° -> the lever-arm criterion is conservative (thesis Sect. 4.5.1). The adjusted ratio drops well below 1 for the outboard case, in line with its much higher simulated φ_max.")
```

## 10  Direct assessment of the required stability (Sect. 4.5.2, Fig. 4.8)

The loading condition is kept, only `GM` is varied (change of `KG`: $GZ\to GZ+\Delta GM\sin\varphi$), and `t_d` = 6.6 s. Below a threshold `GM_cap` the ship capsizes (no flooding is modelled, as in the thesis footnote); the jump in `φ_max` identifies it. The vanishing‑stability angle of the digitised/extrapolated curve is used as capsize criterion.

```python
gms=np.arange(1.0,5.01,0.25); out=[]; t0=time.time()
cfgT=Config(V_thesis,np.pi)
for gm in gms:
    hh=make_hydro(V_thesis,dGM=gm-GM0); d1_=run_phase1(cfgT,hh,7.6)
    _,s2,_=simulate(6.6,cfgT,hh,t_end_after=40.0,dense1=d1_)
    tt_=np.linspace(6.6,46.6,1200); out.append(np.rad2deg(s2.sol(tt_)[3]).max())
out=np.array(out); print(f"{len(gms)} runs in {time.time()-t0:.0f} s")
fig,ax=plt.subplots(figsize=(6.5,3.6)); ax.plot(gms,np.minimum(out,250),'o-'); ax.axvline(GM0,c='gray',ls=':',label='GM of test case (4.49 m)')
ax.set(xlabel='GM [m]',ylabel='φ_max [°]',title='Direct stability assessment (cf. Fig. 4.8)'); ax.legend(); plt.show()
print(pd.DataFrame({'GM [m]':gms,'φ_max [°]':out.round(1)}).to_string(index=False))
safe=gms[out<90]; print(f"\nthreshold stability GM_cap ≈ {safe.min():.2f} m (first GM without capsize on this grid); the test case has GM = {GM0} m")
```

## 11  Two vessels – loss of one rope (new)

The load now hangs from **two vessels**, one rope each, instead of one ship with two cranes. If the rope of one vessel (*released vessel*) fails at t = 0:

* the **remaining vessel** + load is exactly the 11‑DOF system of Sect. 4 (phase 1); when its rope fails after `t_d`, the 6‑DOF ship follows (phase 2);
* the **released vessel** is decoupled from the rest (hydrodynamic interaction and collision are *not* modelled) and is simply **unloaded**: its rope force vanishes at t = 0, so `rhs2` is integrated from its pre‑failure equilibrium (`run_release`).

| | single ship, two cranes (thesis) | two vessels (this section) |
|---|---|---|
| load carried by the ship before the first failure | 100 % | `share` (remaining) / `1 − share` (released) |
| loading condition (`x_G`, `e_G`, displacement) | upright with the full load | `Vessel.design_ballast`: upright under *its own* rope force |
| rope | vertical, tip follows the load position | inclined by `theta` (vessels further apart than the lifting eyes), crane tip fixed on the ship |
| horizontal rope force | none | constant station‑keeping wrench (`F_sk`, `M_sk_z`) – DP/mooring does not react within seconds |
| after the first failure | ship already in equilibrium with the full load | the remaining vessel suddenly carries **100 %** instead of `share`: it sinks and heels *in addition to* the load swing |
| hydrodynamics / hull | one set | one `Vessel` per ship – here both are the thesis hull (**replace with your own data**) |
| released ship | – | unloaded by `1 − share` of the load: rolls and heaves back |

Geometry of the default case: vessels lie side by side with the load between them (`e0 = 270°`: the remaining vessel holds the eye nearest to it, the released vessel the opposite eye, at distance `b = ex·share/(1−share)` from the load COG so that the load is statically balanced). `theta` is the rope inclination.

### 11.1  Set‑up and pre‑failure equilibrium

```python
import pandas as pd
def make_vessel(name): return Vessel(name)      # same hull data as the thesis ship – replace by the data of your own vessels

def two_vessel_case(share=0.5,theta=np.deg2rad(6.),e0=1.5*np.pi,com_rem=com_L,com_rel=None,V_rem=None,V_rel=None):
    """Returns Config of the REMAINING vessel and of the RELEASED vessel (ballast of both designed for the lift).
    share : fraction of the load carried by the remaining vessel before the first failure
    theta : rope inclination of both ropes (vessel spacing larger than the eye spacing)
    e0    : yaw of the load seen from the remaining vessel (270° = side by side, 180° = in tandem with the load alongside)
    com_rem / com_rel : load COG in the ship coordinates of the remaining / released vessel
                        (default: mirror image, load between two vessels lying side by side)"""
    V_rem=V_rem or make_vessel('remaining vessel'); V_rel=V_rel or make_vessel('released vessel')
    b=ex*share/(1-share)                                     # eye of the released vessel on the opposite side of the COG
    if com_rel is None: com_rel=com_rem*np.array([1,-1,1])
    cfg_rem=Config(V_rem,e0,share,theta,design=True,com_L=com_rem)
    cfg_rel=Config(V_rel,e0+np.pi,1-share,theta,design=True,com_L=com_rel,ex=b)
    return cfg_rem,cfg_rel

cfgR,cfgF=two_vessel_case()
rows=[]
for nm,c in [('remaining',cfgR),('released',cfgF)]:
    V=c.V; hh=make_hydro(V); y0=c.y0
    f,n,nu,RES=ship_forces(y0[:6],np.zeros(6),np.zeros(hh.n),hh,y0[2])      # ship forces before the failure
    F_r=c.T_pre*c.u; M_r=np.cross(c.tip-V.r_G,F_r)                          # + force and moment of its rope
    resid=np.abs(np.r_[f+F_r,n+M_r]).max()
    rows.append((nm,c.share,np.rad2deg(c.theta),c.T_pre/g/1e3,V.m_pre/1e3,V.draught_pre,V.GM_pre,V.e_G,V.x_G,V.F_sk[1]/1e3,c.tip[1],resid))
print(pd.DataFrame(rows,columns=['vessel','share','tilt [°]','rope T [t]','displ. [t]','draught [m]','GM [m]','e_G [m]','x_G [m]','F_sk,y [kN]','tip y [m]','residual [N,Nm]']).round(3).to_string(index=False))
print("\nresidual = ship force/moment balance before the first failure (must be ~0: the vessel floats upright and untrimmed).")
```

### 11.4  Quasi‑static check for the remaining vessel (cf. Sect. 9)

After the first failure the remaining vessel carries the **whole** load on its crane, although it was ballasted for `share` of it. The static heel `φc` with the full load, the worst extreme `φx = 2φc − φ1` of the oscillation about it (the vessel is upright before the failure, `φ1 = 0`) and the adjusted area ratio `A2/A1` of the lever‑arm criterion are evaluated with the vessel's own `GZ`, `e_G`, `φ_f`.

```python
def static_heel_V(V,y_tip,z_tip,m_full=m_L):
    # equilibrium of vessel V (displacement m_S+m_full) with the full load hanging from the crane tip (y_tip,z_tip)
    z_t=z_tip-V.KG_S
    def M(p):
        arm=(y_tip+V.e_G)*np.cos(p)-z_t*np.sin(p)
        return -(V.m_S+m_full)*g*(V.GZ(p)[0]-V.e_G*np.cos(p))-m_full*g*arm
    return brentq(M,-0.6,0.6)
def area_ratio_V(V,phi1):
    p2=brentq(V.h2,1e-3,1.0); a=np.linspace(phi1,p2,500); b=np.linspace(p2,V.phi_f,500)
    A1=np.trapezoid(-np.array([V.h2(p) for p in a]),a); A2=np.trapezoid([V.h2(p) for p in b],b); return A1,A2,p2

def quasi_static(c):
    V=c.V; phic=static_heel_V(V,c.tip[1],c.tip[2]); phi1=0.0; phix=2*phic-phi1; phiw=min(phi1,phix) if c.tip[1]>0 else max(phi1,phix)
    A1,A2,p2=area_ratio_V(V,phi1); A1w,A2w,_=area_ratio_V(V,phiw) if c.tip[1]>0 else area_ratio_V(V,-phiw)
    return np.rad2deg(phic),np.rad2deg(phiw),np.rad2deg(p2),A2/A1,A2w/A1w
q=quasi_static(cfgR)
print(pd.DataFrame([('remaining vessel, full load on one crane (share 0.5 -> 1.0)',*q)],
      columns=['case','φc [°]','worst φ at failure [°]','heel after loss φ2 [°]','A2/A1 (φ1=0)','A2/A1 (adjusted)']).round(2).to_string(index=False))
print(f"\nsimulated for the same vessel: φ_max = {RR[:,1].max():.1f}° (worst t_d); downflooding angle φ_f = {np.rad2deg(phi_f):.1f}°")
```

### 11.6  What the two‑vessel layer assumes (and what it does not)

* **Hull data:** both vessels are the thesis hull (`make_vessel`); `GZ_hull(φ)` is kept for the changed displacement, i.e. the effect of the draught change on the lever arm is neglected. Use your own `Vessel(...)` data (and BEM `A_inf`, `B(ω)`) for real vessels.
* **After the first failure the vessels are independent:** no hydrodynamic interaction (12×12 added mass/damping with coupling terms), no collision/contact between the released vessel, its hook and the load.
* **Station keeping:** the horizontal rope force before the failure is held by a constant force/yaw moment at the COG; a real DP/mooring system reacts on a longer time scale.
* **Rope:** rigid and massless as before (no elasticity, no slack/snap load). With long vessel spacing and elastic ropes the first failure also excites a rope/heave transient that is not modelled here.
* **Calm water:** as in the thesis. The pre‑failure state is static, so vessel motions in waves (relative motion of the two vessels, dynamic rope forces before the failure) are not part of this layer – that needs the full coupled two‑vessel model (closed kinematic loop with a rope constraint).
