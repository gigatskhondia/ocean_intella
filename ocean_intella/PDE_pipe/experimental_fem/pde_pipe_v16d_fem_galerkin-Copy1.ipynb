```python
def Πe(φ,θ,ψ):
    return np.array([[np.cos(θ),0,np.cos(φ)*np.sin(θ)],
                  [0,1,-np.sin(φ)],
                  [-np.sin(θ),0,np.cos(φ)*np.cos(θ)]])
```

```python
def Ret_e_t(φ,θ,ψ):
    Cφ=np.array([[1,0,0],
                  [0,np.cos(φ),-np.sin(φ)],
                  [0,np.sin(φ),np.cos(φ)]])

    Cθ=np.array([[np.cos(θ),0,np.sin(θ)],
                  [0,1,0],
                  [-np.sin(θ),0,np.cos(θ)]])

    Cψ=np.array([[np.cos(ψ),-np.sin(ψ),0],
                  [np.sin(ψ),np.cos(ψ),0],
                  [0,0,1]])

    return Cθ @ Cφ @ Cψ
```

```python
diag_J_t_rho = 1e2*np.array([1, 1, 2]) # [m^4] 
J_t_rho = np.diag(diag_J_t_rho)
```

```python
diag_CT = 1e9*np.array([1, 1, 1 ])  
CT=np.diag(diag_CT)
```

```python
diag_CR = 1e11*np.array([1, 1, 1]) 
CR=np.diag(diag_CR)
```

```python
diag_DT = 1.5*np.array([1, 1, 1])
DT=np.diag(diag_DT)
```

```python
diag_DR = 1.5*np.array([1, 1, 1])
DR=np.diag(diag_DR)
```

```python
def I_e_rho(φ,θ,ψ):
    return Ret_e_t(φ,θ,ψ) @ J_t_rho @ Ret_e_t(φ,θ,ψ).T
```

```python
def phi(x,y,z): return np.array([x,y,z]) 
def theta(φ,θ,ψ): return np.array([φ,θ,ψ]) 
```

```python
def ωe(φ,θ,ψ): 
    return Πe(φ,θ,ψ)@theta(φ,θ,ψ)
```

```python
def S(a1, a2, a3 ):
    return np.array([[0, -a3, a2 ],
                         [a3, 0, -a1],
                        [-a2, a1, 0]])
```

```python
def d_s_phi(x,y,z, φ,θ,ψ):
    a1, a2, a3 = ωe(φ,θ,ψ)
    return S(a1, a2, a3 )@phi(x,y,z)
#     return 1/h*(phi(x1,y1,z1)-phi(x2,y2,z2))
```

```python
def ne(x, y, z, φ, θ, ψ):
    return (Ret_e_t(φ,θ,ψ)@CT@Ret_e_t(φ,θ,ψ).T@(d_s_phi(x,y,z, φ, θ, ψ).reshape(3,1)
            -Ret_e_t(φ,θ,ψ)@np.array([1,0,0]).reshape(3,1))).reshape(1,3)
```

```python
def me( φ, θ, ψ):
    return Ret_e_t(φ,θ,ψ)@CR@Ret_e_t(φ,θ,ψ).T@ωe(φ,θ,ψ)
```

```python
def dΠ(φ, θ, ψ, dφ, dθ, dψ):
    return np.array([[-np.sin(θ)*dθ,0,-np.sin(φ)*dφ*np.sin(θ)+np.cos(φ)*np.cos(θ)*dθ],
                  [0,0,-np.cos(φ)*dφ],
                  [-np.cos(θ)*dθ,0,-np.sin(φ)*dφ*np.cos(θ)-np.cos(φ)*np.sin(θ)*dθ]])
```

```python
fe_g = np.array([0,0,-mp*9.81])          
```

```python
def f_t_d(dx,dy,dz): 
    
    vr1 = dx - dv1_curr
    vr2 = dy - dv2_curr  
    vr3 = dz - dv3_curr
    A = np.array([np.abs(vr1) * vr1,
                  np.sqrt(vr2**2 + vr3**2) * vr2,
                 np.sqrt(vr2**2 + vr3**2) * vr3])
    return 0.5 * d0 * qw * (DT@ A)
```

```python
def sigma(x,y,z):
    e3 = np.array([0,0,1])
    
    k = phi(x,y,z) @ e3+d0/2
    
    fe_g2 = np.linalg.norm(fe_g, ord=2)
   
    if k<0:
        k0=0
    elif 0<=k<=d0/20:
        k0=(fe_g2*10*k**2)/((d0/8-d0/40)*(d0))
    else:
        k0=(fe_g2*(k-d0/40))/(d0/8-d0/40)
            
    result = k0 * e3 
   
    return result  
```

```python
# Linear (P1) Galerkin shape functions on the reference element xi in [-1, 1],
# and the 2-point Gauss-Legendre quadrature rule that integrates them exactly.
def N1(xi):
    """Shape function associated with the left (start) node of the element."""
    return 0.5 * (1.0 - xi)

def N2(xi):
    """Shape function associated with the right (end) node of the element."""
    return 0.5 * (1.0 + xi)

# 2-point Gauss-Legendre quadrature points/weights on [-1, 1] (weights are both 1.0)
GAUSS_PTS = (-1.0 / np.sqrt(3), 1.0 / np.sqrt(3))
```

```python
def static_solution(Q):
  
    x, y, z = Q[0:N], Q[2*N:3*N], Q[4*N:5*N]
    dx, dy, dz = Q[1*N:2*N], Q[3*N:4*N], Q[5*N:6*N]
    φ, θ, ψ = Q[6*N:7*N], Q[8*N:9*N], Q[10*N:11*N]
    dφ, dθ, dψ = Q[7*N:8*N], Q[9*N:10*N], Q[11*N:12*N]

    xi_a, xi_b = GAUSS_PTS  # the two Gauss points shared by every element

    # 1. Mesh: N nodes / (N-1) linear elements of length h already defined above.

    # 2. Global stiffness matrix and force vector
    K_global = np.zeros((6 * N, 6 * N))
    F_global = np.zeros(6 * N)

    # 3. Element assembly loop (Galerkin FEM, linear shape functions N1, N2)
    for e in range(1, N):
        node1, node2 = e - 1, e

        # --- Local element stiffness block ---
        # fun1 already carries the N1(xi)*N2(xi)-type weighting for the
        # rotational/translational inertia terms; summing it at the two Gauss
        # points performs the 2-point Gauss quadrature over the element.
        K_element = np.block([
            [fun1(xi_a, φ[node1], θ[node1], ψ[node1]) + fun1(xi_b, φ[node1], θ[node1], ψ[node1]),
             np.zeros((6, 6))],
            [np.zeros((6, 6)),
             fun1(xi_a, φ[node2], θ[node2], ψ[node2]) + fun1(xi_b, φ[node2], θ[node2], ψ[node2])]
        ])

        # --- Local element force vector: weak-form contributions, each
        # already weighted by a linear shape function (N1 or N2) inside
        # fun2..fun6, integrated over the element with 2-point Gauss quadrature ---
        f_internal = 2 * fun2(0, x[node1], y[node1], z[node1], φ[node1], θ[node1], ψ[node1])

        f_gyro_1 = (fun3(xi_a, φ[node1], θ[node1], ψ[node1], dφ[node1], dθ[node1], dψ[node1])
                    + fun3(xi_b, φ[node1], θ[node1], ψ[node1], dφ[node1], dθ[node1], dψ[node1]))

        f_gyro_2 = (fun4(xi_a, φ[node1], θ[node1], ψ[node1], dφ[node1], dθ[node1], dψ[node1])
                    + fun4(xi_b, φ[node1], θ[node1], ψ[node1], dφ[node1], dθ[node1], dψ[node1]))

        f_gravity = fun5(xi_a) + fun5(xi_b)

        f_drag = (fun6(xi_a, φ[node1], θ[node1], ψ[node1], dx[node1], dy[node1], dz[node1],
                       dφ[node1], dθ[node1], dψ[node1])
                  + fun6(xi_b, φ[node1], θ[node1], ψ[node1], dx[node1], dy[node1], dz[node1],
                        dφ[node1], dθ[node1], dψ[node1]))

        F_element = f_internal + f_gyro_1 + f_gyro_2 + f_gravity + f_drag

        # Assemble local element matrix/vector into the global system
        K_global = Assemble(K_global, K_element, node1, node2)
        start, end = 6 * node1, 6 * node1 + 6
        F_global[start:end] += F_element

    K_global[np.isnan(K_global)] = 0
    F_global[np.isnan(F_global)] = 0

    # 4. Enforce Dirichlet boundary condition (node 0 clamped: all 6 DOFs = 0)
    #    via direct row substitution, as in solve_fem_1d.
    pdof = np.array([0, 1, 2, 3, 4, 5])
    fdof = np.setdiff1d(np.arange(6 * N), pdof)

    for i in pdof:
        K_global[i, :] = 0.0
        K_global[i, i] = 1.0
        F_global[i] = 0.0

    # Small regularization on the free DOFs only, for numerical robustness
    # (kept from the original implementation).
#     K_global[np.ix_(fdof, fdof)] += 1e-5 * np.eye(len(fdof))

    # 5. Solve the linear system
    d = np.linalg.solve(K_global, F_global)

    ddx, ddy, ddz, ddφ, ddθ, ddψ = d.reshape(N, 6).T

    ans = np.concatenate([dx, ddx,
                          dy, ddy,
                          dz, ddz,
                          dφ, ddφ,
                          dθ, ddθ,
                          dψ, ddψ,
                          ], axis=0)

    return ans
```

```python
def dynamic_func(t, Q, T):
    
#     print(t)    
    x,y,z=Q[0:N],Q[2*N:3*N],Q[4*N:5*N]
    dx,dy,dz=Q[1*N:2*N],Q[3*N:4*N],Q[5*N:6*N]
    φ,θ,ψ=Q[6*N:7*N],Q[8*N:9*N],Q[10*N:11*N]
    dφ,dθ,dψ=Q[7*N:8*N],Q[9*N:10*N],Q[11*N:12*N]
    
    u_actual = T.force_factor*vessel_motion_func(t)
#     u_actual = [0,0,np.sin(t),0,0,0]
    
    u_actual_x = u_actual[0]
    u_actual_y = u_actual[1]
    u_actual_z = u_actual[2]
    u_actual_phi = u_actual[3]
    u_actual_theta = u_actual[4]
    u_actual_psi = u_actual[5]
        
        
    f2=np.zeros(6*N) 
    f3=np.zeros(6*N) 
    f4=np.zeros(6*N) 
    f5=np.zeros(6*N) 
    f6=np.zeros(6*N) 
    f7=np.zeros(6*N) 
    f8=np.zeros(6*N)
    
    K_stat=np.zeros(shape=(6*N,6*N))
    for j in range(1,N):
        block = np.block([[integrate.fixed_quad(lambda x_int: fun1(x_int,φ[j-1],θ[j-1],ψ[j-1]),-1.0,1.0,n=2)[0],np.zeros((6,6))],
                          [integrate.fixed_quad(lambda x_int: fun1(x_int,φ[j],θ[j],ψ[j]),-1.0,1.0,n=2)[0],np.zeros((6,6))]])
        K_stat = Assemble(K_stat,block,j-1,j)
        start = 6*(j-1)
        end = start + 6
        f2[start:end] += integrate.fixed_quad(lambda x_int:fun2(x_int, x[j-1],y[j-1],z[j-1],φ[j-1],θ[j-1],ψ[j-1],x[j],y[j],z[j],h),-1.0,1.0,n=1)[0]
        f3[start:end] += integrate.fixed_quad(lambda x_int:fun3(x_int, φ[j-1],θ[j-1],ψ[j-1],dφ[j-1],dθ[j-1],dψ[j-1]),-1.0,1.0,n=2)[0]
        f4[start:end] += integrate.fixed_quad(lambda x_int:fun4(x_int, φ[j-1],θ[j-1],ψ[j-1],dφ[j-1],dθ[j-1],dψ[j-1]),-1.0,1.0,n=2)[0]
        f5[start:end] += fun5(-1/np.sqrt(3))+fun5(1/np.sqrt(3))
        f6[start:end] += integrate.fixed_quad(lambda x_int: fun6(x_int,φ[j-1],θ[j-1],ψ[j-1],dx[j-1],dy[j-1],dz[j-1],dφ[j-1], dθ[j-1], dψ[j-1]),-1.0,1.0,n=2)[0]
        f7[start:end] += integrate.fixed_quad(lambda x_int: fun7(x_int,x[j-1],y[j-1],z[j-1]),-1.0,1.0,n=2)[0]
#     for j in range(1, N):
#         block=np.block([[fun1(-1/np.sqrt(3),φ[j-1],θ[j-1],ψ[j-1])+fun1(1/np.sqrt(3),φ[j-1],θ[j-1],ψ[j-1]),np.zeros((6,6))],
#                        [np.zeros((6,6)), fun1(-1/np.sqrt(3),φ[j],θ[j],ψ[j])+fun1(1/np.sqrt(3),φ[j],θ[j],ψ[j])]])
#         K_stat=Assemble(K_stat,block,j-1,j)
#         start = 6*(j-1)
#         end = start + 6
#         f2[start:end] += 2*fun2(0,x[j-1],y[j-1],z[j-1],φ[j-1],θ[j-1],ψ[j-1],x[j],y[j],z[j],h)
#         f3[start:end] += (fun3(-1/np.sqrt(3),φ[j-1],θ[j-1],ψ[j-1],dφ[j-1],dθ[j-1],dψ[j-1])
#                             +fun3(1/np.sqrt(3),φ[j-1],θ[j-1],ψ[j-1],dφ[j-1],dθ[j-1],dψ[j-1]))
#         f4[start:end] +=  (fun4(-1/np.sqrt(3),φ[j-1],θ[j-1],ψ[j-1],dφ[j-1],dθ[j-1],dψ[j-1])
#                             +fun4(1/np.sqrt(3),φ[j-1],θ[j-1],ψ[j-1],dφ[j-1],dθ[j-1],dψ[j-1]))
#         f5[start:end] += fun5(-1/np.sqrt(3))+fun5(1/np.sqrt(3))
#         f6[start:end] += (fun6(-1/np.sqrt(3),φ[j-1],θ[j-1],ψ[j-1],dx[j-1],dy[j-1],dz[j-1],dφ[j-1], dθ[j-1], dψ[j-1])
#                             +fun6(1/np.sqrt(3),φ[j-1],θ[j-1],ψ[j-1],dx[j-1],dy[j-1],dz[j-1],dφ[j-1], dθ[j-1], dψ[j-1]))
#         f7[start:end] += fun7(-1/np.sqrt(3),x[j-1],y[j-1],z[j-1])+fun7(1/np.sqrt(3),x[j-1],y[j-1],z[j-1])
    
    
       
        if t>0.001:
            ax=np.linalg.inv(Ret_e_t(φ[j-1],θ[j-1],ψ[j-1]))@ ne(x[j-1], y[j-1], z[j-1], φ[j-1], θ[j-1], ψ[j-1]).T

            T.top_tension=max(T.top_tension, ax[0][0])            

            ben0=np.linalg.inv(Ret_e_t(φ[j-1],θ[j-1],ψ[j-1])) @ me( φ[j-1], θ[j-1], ψ[j-1]).T
            ben=np.max(ben0[1:])

            I=3.14*(d0**4-dI**4)/64
            strain=np.max(ben)*d0/(2*E*I)    

            T.sagbend_strain=max(T.sagbend_strain, strain)
            
    f8[-6:] += (np.block([[np.identity(3), np.zeros((3,3))], 
                          [np.zeros((3,3)), np.identity(3)]])@u_actual)    
    
    F=(
#         f2
       +f3+f4+f5+f6+f7+f8)  
    
    bc=np.array([[0,1,2,3,4,5], 
                 [0,0,0,0,0,0]]).T
    nd, nd=K_stat.shape
    fdof=np.array([i for i in range(nd)]).T
    d=np.zeros(shape=(len(fdof),))
    Q=np.zeros(shape=(len(fdof),))

    pdof=bc[:,0].astype(int)
    dp=bc[:,1]
    fdof=np.delete(fdof, pdof, 0)
    
    K_stat[np.isnan(K_stat)] = 0
    F[np.isnan(F)] = 0
    
#     s=np.linalg.solve(K_stat[fdof,:][:,fdof], 
#                       (F[fdof].T-np.dot(K_stat[fdof,:][:,pdof],dp.T)).T)
    
#     s=scipy.sparse.linalg.spsolve(scipy.sparse.csr_matrix(K_stat[fdof,:][:,fdof]) ,
#                       (F[fdof].T-np.dot(K_stat[fdof,:][:,pdof],dp.T)).T)

    s=np.linalg.solve(K_stat[fdof,:][:,fdof] + 1e-5 * np.eye(6*(N-1)),
                      (F[fdof].T-np.dot(K_stat[fdof,:][:,pdof],dp.T)).T) 
    
    d[pdof]=dp
    d[fdof]=s.reshape(-1,)
     
    ddx,ddy,ddz,ddφ,ddθ,ddψ=d.reshape(N,6).T
        
        
    ans=np.concatenate([dx, ddx, 
                        dy, ddy,  
                        dz, ddz, 
                        dφ, ddφ,  
                        dθ, ddθ, 
                        dψ, ddψ,  
                       ], axis=0)
    
   
    if t>T.progression[0]:
        T.progression.pop(0)
        print('Physical time: ', t, ' Iteration wall-clock time: ', datetime.now() - T.wall_clock , flush=True)
        T.wall_clock = datetime.now()  
        
        
    T.my_iter+=1 
    
    return ans
```
